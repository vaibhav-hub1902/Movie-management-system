```python
import database
```

```python
USER_CHOICE = """

- a add a movie
- l shows all movies
- s search for movie
- d delete movie
- p update movie rateing
- q quit 

ur choice: """
```

```python
def prompt_insert_movie():

    name = input("enter The movie name:")
    director = input("enter director name:")
    year = int(input("enter the year:"))
    genre = input("enter the genre:")
    rateing = float(input("enter the rateinf"))

    database.insert_movie(name,director,year,genre,rateing)
```

```python
def show_movies():

    movies = database.show_all_movies()

    for movie in movies:
        print(
            f"{movie['name']} directed by {movie['director']}"
            f"Released in {movie['year']}"
            f"{movie['genre']} genre"
            f" rateing {movie['rateing']}"
        )
```

```python
def prompt_search_movie():

    name = input("enter the movie name:")
    movie = database.search_movies(name)

    if movie:
        print("movie found")
        print(f"movie : {movie['name']}")
        print(f"director : {movie['director']}")
        print(f"year : {movie['year']}")
        print(f"genre : {movie['genre']}")
        print(f"rateing : {movie['rateing']}")
    else:
        print("movie not found")
        
```

```python
def prompt_delete_movie():

    name = input("enter the movie name to delete : ")
    database.delete_movie(name)
```

```python
def prompt_update_movie_rateing():

    name = input("enter the movie name to update the rateing:")
    new_rateing = float(input("enter the movie rateing to update:"))

    database.update_movie_rateing(name,new_rateing)
```

```python
def menu():
    
    user_input = input(USER_CHOICE)
    while user_input != 'q':
        if user_input == 'a':
            prompt_insert_movie()
        elif user_input == 'l':
            show_movies()
        elif user_input == 's':
            prompt_search_movie()
        elif user_input == 'd':
            prompt_delete_movie()
        elif user_input == "p":
            prompt_update_movie_rateing()

        user_input = input(USER_CHOICE)
    
```

```python
menu()
```
